```python
import pandas as pd
data=pd.read_csv("data/boston.csv")
data
```

```python
data.MEDV.plot.hist()
```

```python
ax = data.plot.scatter(x='CRIM', y='MEDV', alpha=0.6, s=20, figsize=(8,6))
ax.set_title('CRIM vs MEDV')
ax.set_xlabel('CRIM (per capita crime rate)')
ax.set_ylabel('MEDV (Median value of owner-occupied homes in $1000s)')
ax.grid(True)
```

```python
ax2 = data.plot.scatter(x='AGE', y='MEDV', alpha=0.6, s=20, figsize=(8,6))
ax2.set_title('AGE vs MEDV')
ax2.set_xlabel('AGE')
ax2.set_ylabel('MEDV (Median value of owner-occupied homes in $1000s)')
ax2.grid(True)
```
