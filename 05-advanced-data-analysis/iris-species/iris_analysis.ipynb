```python
import pandas as pd

iris_species_data = pd.read_csv("data/iris.csv")
iris_species_data.info()
```

```python
iris_species_data.plot.scatter(x="petal_length", y="petal_width")
```

```python
color = iris_species_data["species"].map(
    {"setosa": "red", "versicolor": "blue", "virginica": "green"}
)
iris_species_data.plot.scatter(x="petal_length", y="petal_width", c=color)
```

```python
mask = iris_species_data["species"] != "setosa"
# ensure color is defined in this cell (in case the cell that defined it wasn't executed)
color = iris_species_data["species"].map(
	{"setosa": "red", "versicolor": "blue", "virginica": "green"}
)
iris_species_data[mask].plot.scatter(x="petal_length", y="petal_width", c=color[mask])
```
