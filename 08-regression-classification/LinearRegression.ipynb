# Regresión lineal: Advertising example

### Imports y config básica

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

# Para reproducibilidad
np.random.seed(42)

# Opcional: hacer las letras algo más grandes
plt.rcParams["font.size"] = 12
```

### Cargar datos: advertising

```python
# Cargar el fichero Advertising.csv
# Ajusta la ruta si lo tienes en otra carpeta
data = pd.read_csv("Data/Advertising.csv")

# Si tu CSV tiene una primera columna de índice tipo 'Unnamed: 0',
# puedes hacer:
data = pd.read_csv("Data/Advertising.csv", index_col=0)
data.head()

```

### Exploración rápida y scatter Sales vs TV

```python
print("Dimensiones del dataset:", data.shape)
print("\nResumen estadístico:")
display(data.describe())

# Scatter plot Sales vs TV
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(data["TV"], data["sales"], alpha=0.7)
ax.set_xlabel("Inversión en TV (miles de dólares)")
ax.set_ylabel("Ventas (miles de unidades)")
ax.set_title("Relación entre inversión en TV y ventas")
plt.show()
```

### Dividir en train / test y preparar X, Y

Vamos a usar solo TV como predictor, para que las gráficas sean 2D y muy claras.

```python
# Variable explicativa: inversión en TV (reshape para sklearn)
X = data[["TV"]].values  # matriz (n, 1)
y = data["sales"].values # vector (n,)


# División entrenamiento / test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

len(X_train), len(X_test)
```

### Ajustar regresión lineal y mostrar la recta

```python
# Crear y ajustar el modelo de regresión lineal
lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)

# Predicciones en test y cálculo del MSE
y_pred_lin_test = lin_reg.predict(X_test)
mse_lin = mean_squared_error(y_test, y_pred_lin_test)
print(f"MSE de test (Regresión lineal): {mse_lin:.3f}")
```

```python
# Gráfico
fig, ax = plt.subplots(figsize=(7, 5))

ax.scatter(X_train, y_train, alpha=0.6, label="Train")
ax.scatter(X_test, y_test, alpha=0.6, label="Test")

ax.plot(X, lin_reg.predict(X), color="red")

ax.set_xlabel("Inversión en TV (miles de dólares)")
ax.set_ylabel("Ventas (miles de unidades)")
ax.set_title("Ajuste de regresión lineal: Sales ~ TV\n"
             f"MSE test = {mse_lin:.3f}")
ax.legend()
plt.show()
```

```python
# 3D: plano de la regresión lineal usando TV, radio (fijando newspaper en su media)
# Se asume que lin_reg, X, X_train, X_test, y_train, y_test, mse_lin ya existen en el notebook.

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

# Elegimos ejes: TV (x), radio (y) y sales (z). newspaper lo fijamos en su media.
newspaper_mean = X[:, 2].mean()

# Grid para dibujar el plano predicho
tv_vals = np.linspace(X[:, 0].min(), X[:, 0].max(), 40)
radio_vals = np.linspace(X[:, 1].min(), X[:, 1].max(), 40)
TV_grid, RADIO_grid = np.meshgrid(tv_vals, radio_vals)

# Construir la matriz de predicción (flatten luego reshape)
X_grid = np.column_stack([
    TV_grid.ravel(),
    RADIO_grid.ravel(),
    np.full(TV_grid.size, newspaper_mean)
])
sales_pred_grid = lin_reg.predict(X_grid).reshape(TV_grid.shape)

# Superficie (plano aproximado de la regresión lineal en el subespacio TV-radio)
ax.plot_surface(TV_grid, RADIO_grid, sales_pred_grid,
                cmap=plt.cm.Reds, alpha=0.5, linewidth=0, antialiased=True)

# Dibujar los puntos reales (train / test)
ax.scatter(X_train[:, 0], X_train[:, 1], y_train,
           color="C0", label="Train", alpha=0.7)
ax.scatter(X_test[:, 0], X_test[:, 1], y_test,
           color="C1", label="Test", alpha=0.7, marker="^")

ax.set_xlabel("TV (miles $)")
ax.set_ylabel("Radio (miles $)")
ax.set_zlabel("Sales (miles units)")
ax.set_title(f"Regresión lineal (TV, Radio) con newspaper={newspaper_mean:.2f}\nMSE test = {mse_lin:.3f}")
ax.legend()
plt.tight_layout()
plt.show()


# Intentar activar backend interactivo 'widget' solo si ipympl está disponible.
# Si no está disponible, hacemos fallback a 'inline' (sin error).
interactive_backend = False
try:
    import ipympl  # noqa: F401
    # activar widget backend
    get_ipython().run_line_magic("matplotlib", "widget")
    interactive_backend = True
except Exception:
    print("ipympl no disponible: usando backend inline. Para interactividad instale ipympl con: %pip install ipympl")
    get_ipython().run_line_magic("matplotlib", "inline")

# Si hay backend interactivo, cerramos la figura estática y redibujamos en modo widget.
if interactive_backend:
    plt.close(fig)
    fig = plt.figure(figsize=(10, 7))
    ax = fig.add_subplot(111, projection="3d")

    # Volver a dibujar la superficie y los puntos usando las variables ya calculadas
    ax.plot_surface(TV_grid, RADIO_grid, sales_pred_grid,
                    cmap=plt.cm.Reds, alpha=0.5, linewidth=0, antialiased=True)

    ax.scatter(X_train[:, 0], X_train[:, 1], y_train,
               color="C0", label="Train", alpha=0.7)
    ax.scatter(X_test[:, 0], X_test[:, 1], y_test,
               color="C1", label="Test", alpha=0.7, marker="^")

    ax.set_xlabel("TV (miles $)")
    ax.set_ylabel("Radio (miles $)")
    ax.set_zlabel("Sales (miles units)")
    ax.set_title(f"Regresión lineal (TV, Radio) con newspaper={newspaper_mean:.2f}\nMSE test = {mse_lin:.3f}")
    ax.legend()
    plt.show()
else:
    # Ya se mostró la figura estática anteriormente; no hacemos nada más.
    pass
```

# KNN: Advertising example

Ajustar KNN para distintos K y comparar gráficamente

Vamos a probar distintos valores de K y dibujar la curva estimada sobre el mismo scatter.

```python
ks = [1, 5, 30]

fig, axes = plt.subplots(1, len(ks), figsize=(16, 4), sharey=True)

for ax, k in zip(axes, ks):
    knn = KNeighborsRegressor(n_neighbors=k)
    knn.fit(X_train, y_train)

    # Predicciones en test
    tv_grid = np.linspace(X_train[:, 0].min(), X_train[:, 0].max(), 200).reshape(-1, 1)
    y_pred_knn_test = knn.predict(X_test)
    mse_knn = mean_squared_error(y_test, y_pred_knn_test)

    # Predicción sobre el grid para dibujar la curva
    sales_pred_knn_grid = knn.predict(tv_grid)

    # Dibujamos datos y modelo
    ax.scatter(X_train[:, 0], y_train, alpha=0.5, label="Train")
    ax.scatter(X_test[:, 0], y_test, alpha=0.5, label="Test")

    # Curva KNN
    ax.plot(tv_grid.ravel(), sales_pred_knn_grid, linewidth=2,
            label=f"KNN (K={k})")
    ax.set_title(f"K = {k}\nMSE test = {mse_knn:.3f}")
    ax.set_xlabel("TV")
    if ax is axes[0]:
        ax.set_ylabel("Sales")

axes[0].legend()
plt.suptitle("KNN regression: Sales ~ TV para distintos valores de K", y=1.05)
plt.tight_layout()
plt.show()
```

### Comparar cuantitativamente: MSE de test vs K (y regresión lineal)

Aquí hacemos una gráfica de MSE de test en función de K y añadimos una línea horizontal con el MSE de la regresión lineal.

```python
K_values = range(1, 51)
mse_test_knn = []

for k in K_values:
    knn = KNeighborsRegressor(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_pred_knn_test = knn.predict(X_test)
    mse_test_knn.append(mean_squared_error(y_test, y_pred_knn_test))

fig, ax = plt.subplots(figsize=(7, 5))

ax.plot(K_values, mse_test_knn, marker="o", linestyle="-",
        label="KNN (MSE en test)")
ax.axhline(mse_lin, linestyle="--",
           label=f"Regresión lineal (MSE={mse_lin:.3f})")

ax.set_xlabel("K (número de vecinos)")
ax.set_ylabel("MSE de test")
ax.set_title("Comparación de rendimiento: KNN vs Regresión lineal")
ax.legend()
plt.show()
```
